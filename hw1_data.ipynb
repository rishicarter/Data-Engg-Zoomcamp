```python
zone = pd.read_csv("https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc/taxi_zone_lookup.csv")
```

```python
zone.head()
```

```python
df = pd.merge(df, zone, how="inner", right_on='LocationID', left_on='PULocationID')
df.head()
```

```python
df_nov18 = df[(df['lpep_pickup_datetime'].dt.month == 11) & (df['lpep_pickup_datetime'].dt.day == 18)]
```

```python
df_nov18.shape
```

```python
df_nov18.groupby('Zone').agg({"total_amount": "sum"}).reset_index().sort_values(by="total_amount", ascending=False)
```

```python
df['drop_zone'] = df.set_index(["DOLocationID"]).map(zone.set_index(["LocationID"])['Zone'])
```
